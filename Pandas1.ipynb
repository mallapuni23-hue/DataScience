# Day 1

```python
import pandas as pd
age = pd.Series([10,20,30,40], index = ["Ram", "Sita", "Chandu", "Dikshya"])
print(age)
print(age["Ram"])
```

# Series from Dictionary

```python
marks = {
    "Ram" : 89,
    "Sita" : 88,
    "Hari" : 78
}
s = pd.Series(marks)
print(s)
print(s["Sita"])
```

# Series Operations

```python
marks = pd.Series([99,98,100,78,89])
print(marks.mean())
print(marks.max())
print(marks.min())
print(marks.sum())
print(marks.count())
print(marks.std())
```

```python
print(marks.describe())
```

# Data Frame

```python
data = {
    "name" : ["Ram", "Sita", "Hari", "Chandu"],
    "age" : [20,21,22,34],
    "marks" : [56,67,89,80]
}
df = pd.DataFrame(data)
print(df)
```
