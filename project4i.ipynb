# Data visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('titanicclean.csv')

print(df.head())
```

```python
df.info()
```

# Survival count

```python
sns.countplot(x='survived', data=df)
plt.show()
```

# embarked

```python
explode=[0,0,0.1]
df['embark_town'].value_counts().plot.pie(
    autopct='%1.1f%%',
    startangle=90,
    colors = ['#ff9999', '#66b3ff', '#99ff99',],explode=explode
)

plt.title('Passengers by Embarkation')
plt.ylabel('')
plt.show()
```

# Line graph - Parch

```python
parch_count = df['parch'].value_counts().sort_index()

plt.plot(parch_count.index, parch_count.values, marker='o')

plt.title('Passengers by Number of Parents/Children')
plt.xlabel('Number of Parents/Children')
plt.ylabel('Number of Passengers')
plt.show()
```

# Age distribution

```python
sns.histplot(data=df, x='age', bins=20)
plt.title('Distribution of Passenger Age')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
# total no of person at that age
plt.show()
```

# Fare distribution

```python
sns.histplot(data=df, x='fare', bins=20)
plt.title('Distribution of Ticket Fare')
plt.xlabel('Fare')
plt.ylabel('Number of Passengers')
plt.show()
```

# Male vs Female

```python
sns.countplot(x='sex', data=df)
plt.title('Passengers by Sex')
plt.xlabel('Sex')
plt.ylabel('Number of Passengers')
plt.show()
```

# passenger class

```python
plt.figure(figsize=(10,6))
sns.countplot(x='pclass', data=df)
plt.title('Passengers by Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')
plt.show()
```

# survival age

```python
sns.boxplot(x='survived', y='age', data=df)
plt.title('Age Distribution by Survival')
plt.xlabel('Survived')
plt.ylabel('Age')
plt.show()
```

# pclass survival

```python
sns.violinplot(x='pclass', y='age', data=df)

plt.title('Age distribution by Passenger Class ')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.show()
```

```python
sns.histplot(
    data = df,
    x = "age",#fare,parch,sibsps
    kde=True
)

plt.show()
```

# Correlation

```python
correlation = df.corr(numeric_only=True)

print(correlation)
```

# Heatmap

```python
sns.heatmap(correlation,annot=True)
# sns.heatmap(correlation) //annot = True shows value 
# a heatmap is a color-coded chart that makes it easy 
# to see patterns, trends, and areas of high or low activity at a glance.

plt.show()
```

# pairplot

```python
# plt.Figure(figsize=(10,6))
sns.pairplot(
    df,
    hue = 'survived')# A collection of plots that shows relationships between multiple variables in a dataset.
plt.show()
```
