# Missing data handling

```python
data = {
    "Name" : ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Age" : [25,None,35,None,45],
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
# 1----> filling missing values with mean
df["Age"] = df['Age'].fillna(df['Age'].mean())
print(df)
```

```python
# 2----> dropping missing values
print(df.dropna())
```

# Data transformation

```python
data = {
    "Name" : ["Alice", "Bob", "Charlie", "David", "Eve"],
    "marks" : [80,90,75,85,95],
}
df = pd.DataFrame(data)
print(df)
```

```python
# add percentage column
df['percentage'] = df['marks']/100*100
print(df)
```

```python
# increment marks by 5%
df['marks'] = df['marks']*1.05
print(df)
```

# GroupBy

```python
data = {
    'dept' : ['HR', 'IT','Finance', 'Marketing', 'Sales'],
    'salary' : [50000,75000,60000,65000,70000]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.describe())
```

```python
print(df.groupby('dept')['salary'].mean())
```

```python
import pandas as pd
data = {
    'Name' : ['a', 'b', 'c', 'd', 'e'],
    'age' : [25,30,35,40,45],
    'gender' : ['F','M','M','M','F']
}
df = pd.DataFrame(data)
print(df)
```

```python
df.groupby('gender')['age'].mean()
```

```python
df.groupby('gender')['age'].sum()
```

```python
df.groupby('gender')['age'].count()
```

# Sorting

```python
data = {
    "Name" : ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Age" : [25,30,35,40,45],
    "City" : ['New York', 'Los Angeles', "Chicago", 'Houston', 'Phoenix'],
    "marks" : [80,90,75,85,95]
}
df = pd.DataFrame(data)
print(df)
```

```python
# sort by ascendin order of marks
print(df.sort_values('marks', ascending = True))
```
