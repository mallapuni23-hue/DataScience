```python
import pandas as pd
df = pd.read_csv("employee_data_dirty.csv")
print(df)
```

```python
df['Name'] = df['Name'].str.strip()
df = df.drop_duplicates(subset=['Name']).reset_index(drop=True)
print(df)
```

```python
print(df.isnull())
```

```python
df.info()
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
print(df)
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
print(df)
```

```python
df.loc[df['Salary'] < 0, 'Salary'] = None

df['Salary'] = df['Salary'].fillna(df['Salary'].mean())

print(df)
```

```python
df['City'] = df['City'].str.title()
print(df)
```

```python
df['City'] = df['City'].str.strip()
print(df)
```

```python
df['Gender'] = df['Gender'].str.title()
print(df)
```

```python
df = df[df['Email'].str.contains('@',na=False)]
df = df.dropna()
print(df)
```

```python
df['Age'] = df['Age'].round().astype(float)
pd.set_option('display.float_format', '{:.6f}'.format)
print(df)
```

```python
df['Join_Date'] = pd.to_datetime(
    df['Join_Date'],
    errors='coerce',
    format='mixed'
)

df['Join_Date'] = df['Join_Date'].dt.strftime('%d-%m-%Y')
print(df)
```

```python
df.to_csv("clean_employees_data.csv",index = True)
```
