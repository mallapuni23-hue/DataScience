```python
import pandas as pd
df = pd.read_csv("students_messy.csv")
print(df)
```

```python
df.to_csv("copy_students_messy.csv", index = True)
```

```python
df.info()
```

```python
df = df.drop_duplicates(subset=['Name']).reset_index(drop=True)
print(df)
```

```python
df['Age'] = df['Age'].fillna (df['Age'].mean())
print(df)
```

```python
df["Marks"] = df['Marks'].fillna(df['Marks'].mean())
print(df)
```

```python
df = df.dropna().reset_index(drop=True)
print(df)
```

```python
df['City'] = df['City'].str.strip()
print(df)
```

```python
df['City'] = df['City'].str.lower().replace({
    'kathmandu' : 'Kathmandu',
    'lalitpur' : 'Lalitpur',
    'pokhara' : 'Pokhara',
    'butwal' : 'Butwal',
})
print(df)
```

```python
df['Gender'] = df['Gender'].replace({
    "Male" : "M",
    "Female" : "F",
    'male' : 'M',
})
print(df)
```

```python
# Convert invalid dates to NaT
df['Join_Date'] = pd.to_datetime(df['Join_Date'], errors='coerce')

# Remove rows where Join_Date is invalid/missing
df = df.dropna(subset=['Join_Date']).reset_index(drop=True)

print(df)
```

```python
# Make negative age null
df.loc[df['Age'] < 0, 'Age'] = None

# Fill null age with mean
df['Age'] = df['Age'].fillna(df['Age'].mean())

print(df)
```

```python
df['Name'] = df['Name'].str.strip()
print(df)
```

```python
df['Name'] = df['Name'].replace({
    'prabin': 'Prabin',
})
print(df)
```

```python
df['Age'] = df['Age'].round().astype(float)
pd.set_option('display.float_format', '{:.6f}'.format)
print(df)
```

```python
df.to_csv('Cleaned_Stud_data.csv',index=True)
```
