# Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("iris")
sns.countplot(data=df, x="species")
plt.title("Species Distribution")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")
df ["species"].unique()
```

```python
# there is 50/50 species
sns.histplot(
    data = df,
    x = "petal_width",
    kde=True
)

plt.show()
```

```python
sns.boxplot(
    data =df,
    x = "species",
    y = "sepal_length"
)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

sns.scatterplot(
    data = df,
    x = "sepal_length",
    y = "sepal_width"
)

plt.show()

```

```python
sns.scatterplot(
    data = df,
    x = "petal_length",
    y = "petal_width"

)

plt.title("Petal length vs Petal width")
plt.xlabel("Petal length")
plt.ylabel("Petal width")


plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.countplot(
    data = df,
    x = "petal_length"
)
plt.show()
```

# Violin Plot

```python
import seaborn as sns
df = sns.load_dataset("iris")
import pandas as pd
import matplotlib.pyplot as plt
sns.violinplot(
    data = df,
    x = "species",
    y = "petal_length"
)
plt.show()
```

Correlation

```python
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("iris")
# Correlation tells us how two numerical variables are related to each other.
correlation = df.corr(numeric_only=True)

print(correlation)
```

```python
sns.heatmap(correlation,annot=True)
# sns.heatmap(correlation) //annot = True shows value 

plt.show()
```
