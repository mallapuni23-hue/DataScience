```python
import numpy as np
import matplotlib.pyplot as plt

#Generate 500 random students marks (normally distributed around 72)
np.random.seed(42)
marks = np.random.normal(loc = 72, scale = 12, size = 500). clip(40,100)

#define 10 - point bin boundaries: [40,50,60,70,80,90,100]
bin_edges = range(40,105,10)

plt.figure(figsize=(9,5))

# bins = bin_edges determines where the bars are drawn and data is grouped
plt.hist(marks,bins=bin_edges,color='skyblue', edgecolor ='black')
# plt.xticks(bin_edges) lines up the axis numbers exactly with bar boundaries.
plt.xticks(bin_edges)
plt.title('Final Exam Score Distribution (N=500)')
plt.xlabel('Score Range')
plt.ylabel('Number of Students')
plt.grid(axis='y', linestyle = '--')

plt.show()
```

# Scatter Plot

```python
hours=[1,2,3,4,5,6]
marks=[50,55,60,65,70,75]

plt.scatter(hours,marks,color='purple',marker='o')

plt.title('Hours Studied vs Marks.Obtained')
plt.xlabel('Hours Studied')
plt.ylabel('Marks Obtained')

plt.show()
```

# Pie-Chart

```python
subjects=['Maths','Science','English','History','Art']
hours_studied=[5,4,6,3,2]

plt.pie(hours_studied,labels=subjects,autopct='%1.1f%%',startangle=90,colors=['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0']
)
plt.title("Hours Studied by Students")
plt.show()
```

```python
subjects=['Maths','Science','English','History','Art']
hours_studied=[5,4,6,3,2]

explode=[0.1,0,0,0,0]

plt.figure(figsize=(8,8))
plt.pie(hours_studied,labels=subjects,autopct='%1.1f%%',startangle=90,colors=['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0'], explode=explode
)
plt.title("Hours Studied by Students")
plt.show()
```
