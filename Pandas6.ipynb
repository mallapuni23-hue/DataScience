# Seaborn Scatter plot

```python
import seaborn as sns
import matplotlib.pyplot as plt


df = sns.load_dataset("tips")

sns.scatterplot(
    data = df,
    x = "total_bill",
    y = "tip"

)

plt.show()
```

```python
sns.scatterplot(
    data = df,
    x = "total_bill",
    y = "tip",
    hue = "sex"
)
plt.show()
```

# Seaborn Line plot

```python
sns.lineplot(
    data = df,
    x = "size",
    y = "tip"
)

plt.show()
```

```python
sns.barplot(
    data = df,
    x = "day",
    y = "total_bill",
    errorbar=None
)

plt.show()
```

# Count plot

```python
# A count plot is useful when we want to know how many observations belong to each category
# how many records are from:
# thursday?
# friday?
# saturday?
# sunday?
sns.countplot(
    data=df,
    x = "day"
)
plt.show()
```

# Box plot with seaborn

# heatmap

```python
# A heatmap reprsents values using intensities.
# One of the most important applications in Data Science in correlation visualization.
```

```python
# first calculate correlation:
correlation = df.corr(numeric_only=True)


print(correlation)
```

```python
sns.heatmap(correlation)

plt.show()
```

```python
sns.heatmap(
    correlation,
    annot=True
)
plt.show()
# annot = true displays the numerical values inside the cells.
```
